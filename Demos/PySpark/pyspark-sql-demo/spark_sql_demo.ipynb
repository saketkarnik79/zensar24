```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Spark SQL Employee Analytics") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Version:", spark.version)
```

```python
employees_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("data/employees.csv")
employees_df.show(truncate=False)
```

```python
departments_df = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("data/departments.csv")

departments_df.show()
```

```python
departments_df.createOrReplaceTempView("departments")
```

```python
spark.sql("SHOW TABLES").show()
```

```python
spark.sql("""
    SELECT
        e.emp_id,
        e.name,
        e.department,
        d.manager,
        d.department_location,
        e.salary
    FROM employees e
    INNER JOIN departments d
        ON e.department = d.department
""").show()
```

```python
spark.sql("""
    SELECT
        e.emp_id,
        e.name,
        e.department,
        d.manager,
        d.department_location,
        e.salary
    FROM employees e
    LEFT JOIN departments d
        ON e.department = d.department
""").show()
```

```python
spark.sql("""
    SELECT
        e.emp_id,
        e.name,
        e.department,
        d.manager,
        d.department_location,
        e.salary
    FROM employees e
    RIGHT JOIN departments d
        ON e.department = d.department
""").show()
```

```python
spark.sql("""
    SELECT
        e.emp_id,
        e.name,
        e.department,
        d.manager,
        d.department_location,
        e.salary
    FROM employees e
    FULL JOIN departments d
        ON e.department = d.department
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        salary,
        ROW_NUMBER() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS salary_row_number
    FROM employees
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        salary,
        RANK() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS salary_rank
    FROM employees
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        salary,
        DENSE_RANK() OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS salary_rank
    FROM employees
""").show()
```

```python
top_employees = spark.sql("""
    SELECT *
    FROM (
        SELECT
            emp_id,
            name,
            department,
            salary,
            ROW_NUMBER() OVER (
                PARTITION BY department
                ORDER BY salary DESC
            ) AS row_num
        FROM employees
    )
    WHERE row_num <= 2
""")

top_employees.show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        salary,
        LAG(salary) OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS previous_salary
    FROM employees
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        salary,
        LEAD(salary) OVER (
            PARTITION BY department
            ORDER BY salary DESC
        ) AS next_salary
    FROM employees
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        joining_date,
        QUARTER(joining_date) AS joining_quarter
    FROM employees
""").show()
```

```python
hiring_trend = spark.sql("""
    SELECT
        YEAR(joining_date) AS joining_year,
        COUNT(*) AS employees_joined
    FROM employees
    GROUP BY YEAR(joining_date)
    ORDER BY joining_year
""")

hiring_trend.show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        department,
        joining_date
    FROM employees
    WHERE YEAR(joining_date) = 2024
    ORDER BY joining_date
""").show()
```

```python
spark.sql("""
    SELECT
        emp_id,
        name,
        joining_date,
        DATEDIFF(CURRENT_DATE(), joining_date) AS tenure_days
    FROM employees
    ORDER BY tenure_days DESC
""").show()
```
